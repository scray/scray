```python
from scray.client import ScrayClient
from scray.client.config import ScrayClientConfig
from scray.client.models.versioned_data import VersionedData
import json

config = ScrayClientConfig(
    host_address = "http://ml-integration.research.dev.seeburger.de",
    port = 8082
)
client = ScrayClient(client_config=config)
```

```python
vd1 = VersionedData(
    data_source="s1",
    merge_key="_",
    version=0,
    data=json.dumps({
        "processingEnv": "http://scray.org/ai/test/env/see/os/k8s",
        "state": "RUNNING"
    }),
    version_key=0
)

vd2 = VersionedData(
    data_source="s2",
    merge_key="_",
    version=0,
    data=json.dumps({
        "processingEnv": "http://scray.org/ai/test/env/see/os/k8s",
        "state": "RUNNING"
    }),
    version_key=0
)

client.updateVersion(vd1)
client.updateVersion(vd2)
```
